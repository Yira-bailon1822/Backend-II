Estrategia terrestre

```python
class TerrestrialShippingStrategy:
    def calculate_cost(self, weight_in_kg: float, distance_in_km: float) -> float:
        # Tarifa base $50 + $5 por cada kg + $1 por cada km
        return 50.0 + (weight_in_kg * 5.0) + (distance_in_km * 1.0)
```

Estrategia Aerea

```python
class AerialShippingStrategy:
    def calculate_cost(self, weight_in_kg: float, distance_in_km: float) -> float:
        # Tarifa base $150 + $20 por cada kg + $5 por cada km
        return 150.0 + (weight_in_kg * 20.0) + (distance_in_km * 5.0)
```

 Calculador de peso para dron

```python
class DroneWeightCalculator:
    MAX_BASE_WEIGHT_KG = 2.0
    EXTRA_FEE_PER_KG = 50.0

    def calculate_extra_weight_cost(self, weight_in_kg: float) -> float:
        if weight_in_kg > self.MAX_BASE_WEIGHT_KG:
            return (weight_in_kg - self.MAX_BASE_WEIGHT_KG) * self.EXTRA_FEE_PER_KG
        return 0.0
```

Estrategia Dron Express

```python
class DroneExpressShippingStrategy:
    def __init__(self):
        self._weight_calculator = DroneWeightCalculator()

    def calculate_cost(self, weight_in_kg: float, distance_in_km: float) -> float:
        base_cost = 300.0
        # La distancia no importa; delegamos el cálculo del peso extra al especialista
        extra_weight_cost = self._weight_calculator.calculate_extra_weight_cost(weight_in_kg)
        return base_cost + extra_weight_cost
```

```python
class Order:
    def __init__(self, weight_in_kg: float, distance_in_km: float, shipping_strategy=None):
        self.weight_in_kg = weight_in_kg
        self.distance_in_km = distance_in_km
        # Por defecto usamos la estrategia terrestre si no se especifica otra
        self.shipping_strategy = shipping_strategy if shipping_strategy is not None else TerrestrialShippingStrategy()

    def calculate_shipping_cost(self) -> float:
        if self.shipping_strategy is None:
            raise ValueError("Debe asignar un método de envío válido.")

        return self.shipping_strategy.calculate_cost(self.weight_in_kg, self.distance_in_km)
```
